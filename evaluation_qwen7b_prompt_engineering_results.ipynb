### Загрузка данных с Гугл диска

```python
from google.colab import drive
drive.mount('/content/drive')
import pandas as pd
res_h2 = pd.read_csv('/content/drive/MyDrive/ПЗАД проект/Evaluation/PE_new_h2.csv')
res_h1 = pd.read_csv('/content/drive/MyDrive/ПЗАД проект/Evaluation/PE_new_h1.csv')
res_base = pd.read_csv('/content/drive/MyDrive/ПЗАД проект/Evaluation/PE_new_base.csv')
res_h3 = pd.read_csv('/content/drive/MyDrive/ПЗАД проект/Evaluation/PE_new_h3.csv')
```

### Вывод средних оценок

```python
print('Метрики промпт-инжиниринга\n')
print(f'Base\nBERT: {res_base['reward_bert'].mean():.3f}\nLLM: {res_base['reward_llm'].mean():.3f}\nTotal: {res_base['reward_total'].mean():.3f}\n')
print(f'H1\nBERT: {res_h1['reward_bert'].mean():.3f}\nLLM: {res_h1['reward_llm'].mean():.3f}\nTotal: {res_h1['reward_total'].mean():.3f}\n')
print(f'H2\nBERT: {res_h2['reward_bert'].mean():.3f}\nLLM: {res_h2['reward_llm'].mean():.3f}\nTotal: {res_h2['reward_total'].mean():.3f}\n')
print(f'H3\nBERT: {res_h3['reward_bert'].mean():.3f}\nLLM: {res_h3['reward_llm'].mean():.3f}\nTotal: {res_h3['reward_total'].mean():.3f}\n')
```

### Проверим статзначимый аплифт для 3 гипотез с контролем FPR$\le$0.05

```python
from statsmodels.stats.weightstats import ttest_ind

_, p1, _ = ttest_ind(res_base['reward_total'], res_h1['reward_total'], alternative='smaller')
_, p2, _ = ttest_ind(res_base['reward_total'], res_h2['reward_total'], alternative='smaller')
_, p3, _ = ttest_ind(res_base['reward_total'], res_h3['reward_total'], alternative='smaller')

print(f'p1: {p1:.3f}\np2: {p2:.3f}\np3: {p3:.3f}\nalpha Bonferroni: {0.05/3:.3f}')
```
